# Tutorial 1: Understanding a Perceptron
Steps 1–4 (the Perceptron class) are in `Perceptron.py`. This notebook loads the Iris data, trains and tests the perceptron, and completes **Task 1** and **Task 3**.

## Step 1–4: Import the Perceptron class

```python
from Perceptron import Perceptron, PerceptronBinary
```

## Step 5: Loading Data

```python
import pandas as pd                    # data manipulation and analysis
from sklearn.utils import shuffle      # to shuffle the data
import numpy as np                     # numerical operations

# The dataset has no header row, so header=None
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header=None)

# Shuffle rows to randomize the order of the data
df = shuffle(df)
df.head()
```

### Separating the data (X) from the labels (y)

```python
# First four columns: sepal length, sepal width, petal length, petal width
X = df.iloc[:, 0:4].values

# Fifth column: species of the flower
y = df.iloc[:, 4].values

print(X[0:5])
print(y[0:5])
```

### Splitting the data into train and test set

```python
from sklearn.model_selection import train_test_split

# 75% for train and 25% for test
train_data, test_data, train_labels, test_labels = train_test_split(X, y, test_size=0.25)

# Encoding the labels: 1 for setosa, -1 otherwise
train_labels = np.where(train_labels == 'Iris-setosa', 1, -1)
test_labels = np.where(test_labels == 'Iris-setosa', 1, -1)

print('Train data:', train_data[0:2])
print('Train labels:', train_labels[0:2])
print('Test data:', test_data[0:2])
print('Test labels:', test_labels[0:2])
```

## Step 6: Training the perceptron

```python
from sklearn.linear_model import Perceptron as SkPerceptron
# (imported as SkPerceptron so it does not clash with our own Perceptron class)

perceptron = SkPerceptron(eta0=0.1, max_iter=10)
# eta0=0.1   : learning rate
# max_iter=10: maximum number of iterations (epochs)

perceptron.fit(train_data, train_labels)
```

## Step 7: Making predictions on test data

```python
test_preds = perceptron.predict(test_data)
# 1 for 'Iris-setosa', -1 for other species
print(test_preds)
```

## Step 8: Measuring Performance

```python
from sklearn.metrics import accuracy_score

y_preds = perceptron.predict(test_data)
accuracy = accuracy_score(y_preds, test_labels)
print('Accuracy:', round(accuracy, 2) * 100, '%')
```

## Task 1: Predict from manual input
Enter the four flower measurements (in cm). The trained model predicts whether the flower is Iris-setosa or not.

*Try:* `5.1, 3.5, 1.4, 0.2` (setosa) or `6.7, 3.0, 5.2, 2.3` (not setosa).

```python
sepal_length = float(input("Enter sepal length (cm): "))
sepal_width  = float(input("Enter sepal width (cm): "))
petal_length = float(input("Enter petal length (cm): "))
petal_width  = float(input("Enter petal width (cm): "))

# The model expects a 2D array: one row with four features
sample = np.array([[sepal_length, sepal_width, petal_length, petal_width]])
prediction = perceptron.predict(sample)[0]

print("\nInput features:", sample[0])
if prediction == 1:
    print("Prediction: Iris-setosa")
else:
    print("Prediction: Not Iris-setosa")
```

## Task 3: Replace the labels 1, -1 with 1 & 0
Setosa is now encoded as **1** and the other species as **0**.

For our own perceptron, the step function in `predict()` must also output 0 instead of -1, otherwise the prediction could never match a 0 label. That is done in the `PerceptronBinary` class in `Perceptron.py`. The update rule η(y − ŷ)·x still works, because y − ŷ is now 1, 0 or −1.

```python
# Re-encode the labels: 1 for setosa, 0 otherwise
train_labels_01 = np.where(train_labels == 1, 1, 0)
test_labels_01 = np.where(test_labels == 1, 1, 0)

print('Train labels:', train_labels_01[0:10])
print('Test labels:', test_labels_01[0:10])
```

### 3a. scikit-learn Perceptron with 1/0 labels

```python
perceptron_01 = SkPerceptron(eta0=0.1, max_iter=10)
perceptron_01.fit(train_data, train_labels_01)

preds_01 = perceptron_01.predict(test_data)
print('Predictions:', preds_01)
print('Accuracy:', round(accuracy_score(preds_01, test_labels_01), 2) * 100, '%')
```

### 3b. Our own perceptron (from `Perceptron.py`) with 1/0 labels
Note: the class prints the weights after every update, so this cell gives a long output.

```python
my_perceptron = PerceptronBinary(eta=0.01, n_iter=10)
my_perceptron.fit(train_data, train_labels_01)
```

```python
my_preds = my_perceptron.predict(test_data)
print('Predictions:', my_preds)
print('Errors in each iteration:', my_perceptron.errors_)
print('Accuracy:', round(accuracy_score(my_preds, test_labels_01), 2) * 100, '%')
```
